```python
# ============================================================
# CELL 1 - IMPORT LIBRARY
# ============================================================

# NumPy digunakan untuk membuat dan mengolah data dalam bentuk array
import numpy as np

# Pandas digunakan untuk menampilkan dan mengolah data dalam bentuk tabel
import pandas as pd

# linkage dan dendrogram digunakan untuk melakukan
# Hierarchical Clustering dan membuat visualisasi dendrogram
from scipy.cluster.hierarchy import linkage, dendrogram

# pdist digunakan untuk menghitung jarak antar data
# squareform digunakan untuk mengubah hasil jarak menjadi bentuk matriks
from scipy.spatial.distance import pdist, squareform

# Matplotlib digunakan untuk membuat visualisasi
import matplotlib.pyplot as plt
```

```python
# ============================================================
# CELL 2 - DATASET SEDERHANA
# ============================================================

# Membuat dataset sederhana yang terdiri dari 6 titik data
# Setiap baris merupakan satu titik data dengan koordinat (X, Y)

data = np.array([
    [1.0, 1.0],   # a
    [1.5, 1.5],   # b
    [5.0, 5.0],   # c
    [3.0, 4.0],   # d
    [4.0, 4.0],   # e
    [3.0, 3.5]    # f
])

# Memberikan label pada setiap titik data
labels = ['a', 'b', 'c', 'd', 'e', 'f']

# Menampilkan dataset dalam bentuk tabel
data_df = pd.DataFrame(
    data,
    columns=['X', 'Y'],
    index=labels
)

print("=== Dataset Sederhana ===")
display(data_df)
```

```python
# ============================================================
# CELL 3 - MENGHITUNG EUCLIDEAN DISTANCE
# ============================================================

# Menghitung jarak Euclidean antara setiap pasangan titik data
distance_matrix = pdist(data, metric='euclidean')

# Mengubah hasil perhitungan jarak menjadi bentuk matriks
distance_matrix_square = squareform(distance_matrix)

# Membuat DataFrame agar distance matrix lebih mudah dibaca
distance_df = pd.DataFrame(
    distance_matrix_square,
    columns=labels,
    index=labels
)

print("=== Distance Matrix (Euclidean) ===")
display(distance_df.round(6))
```

```python
# ============================================================
# CELL 4 - HIERARCHICAL CLUSTERING: SINGLE LINKAGE
# ============================================================

# Membuat hierarchical clustering menggunakan metode single linkage
#
# Single linkage menggunakan jarak TERDEKAT antara dua cluster.
#
# Metric euclidean digunakan untuk menghitung jarak antar data.

linked_single = linkage(
    data,
    method='single',
    metric='euclidean'
)

# Menampilkan hasil proses clustering
print("=== Hasil Single Linkage ===")
print(linked_single)
```

```python
# ============================================================
# CELL 5 - DENDROGRAM SINGLE LINKAGE
# ============================================================

# Membuat ukuran gambar agar dendrogram mudah dibaca
plt.figure(figsize=(12, 8))

# Membuat dendrogram berdasarkan hasil single linkage
dendrogram(
    linked_single,
    labels=labels,
    leaf_rotation=90
)

# Memberikan judul pada grafik
plt.title("Hierarchical Clustering Dendrogram - Single Linkage")

# Memberikan nama sumbu X
plt.xlabel("Data Points")

# Memberikan nama sumbu Y
plt.ylabel("Distance")

# Menampilkan grid untuk membantu membaca nilai distance
plt.grid(True)

# Menampilkan grafik
plt.show()
```

```python
# ============================================================
# CELL 6 - HIERARCHICAL CLUSTERING: COMPLETE LINKAGE
# ============================================================

# Membuat hierarchical clustering menggunakan metode complete linkage
#
# Complete linkage menggunakan jarak TERJAUH antara anggota
# dari dua cluster ketika menentukan jarak antar-cluster.

linked_complete = linkage(
    data,
    method='complete',
    metric='euclidean'
)

# Menampilkan hasil clustering
print("=== Hasil Complete Linkage ===")
print(linked_complete)
```

```python
# ============================================================
# CELL 7 - DENDROGRAM COMPLETE LINKAGE
# ============================================================

# Membuat ukuran gambar
plt.figure(figsize=(12, 8))

# Membuat dendrogram menggunakan hasil complete linkage
dendrogram(
    linked_complete,
    labels=labels,
    leaf_rotation=90
)

# Memberikan judul grafik
plt.title("Hierarchical Clustering Dendrogram - Complete Linkage")

# Memberikan nama sumbu X dan Y
plt.xlabel("Data Points")
plt.ylabel("Distance")

# Menampilkan grid
plt.grid(True)

# Menampilkan grafik
plt.show()
```

```python
# ============================================================
# CELL 8 - HIERARCHICAL CLUSTERING: AVERAGE LINKAGE
# ============================================================

# Membuat hierarchical clustering menggunakan metode average linkage
#
# Average linkage menggunakan rata-rata jarak antara
# anggota dari dua cluster.

linked_average = linkage(
    data,
    method='average',
    metric='euclidean'
)

# Menampilkan hasil clustering
print("=== Hasil Average Linkage ===")
print(linked_average)
```

```python
# ============================================================
# CELL 9 - DENDROGRAM AVERAGE LINKAGE
# ============================================================

# Membuat ukuran gambar
plt.figure(figsize=(12, 8))

# Membuat dendrogram menggunakan hasil average linkage
dendrogram(
    linked_average,
    labels=labels,
    leaf_rotation=90
)

# Memberikan judul grafik
plt.title("Hierarchical Clustering Dendrogram - Average Linkage")

# Memberikan nama sumbu
plt.xlabel("Data Points")
plt.ylabel("Distance")

# Menampilkan grid
plt.grid(True)

# Menampilkan grafik
plt.show()
```

```python
# ============================================================
# CELL 10 - PERBANDINGAN SINGLE, COMPLETE, DAN AVERAGE
# ============================================================

# Menyiapkan tiga metode linkage yang akan dibandingkan
methods = {
    "Single Linkage": linked_single,
    "Complete Linkage": linked_complete,
    "Average Linkage": linked_average
}

# Membuat tiga grafik secara terpisah
for method_name, linkage_result in methods.items():

    # Membuat ukuran grafik
    plt.figure(figsize=(12, 8))

    # Membuat dendrogram
    dendrogram(
        linkage_result,
        labels=labels,
        leaf_rotation=90
    )

    # Memberikan judul sesuai metode
    plt.title(
        f"Hierarchical Clustering Dendrogram - {method_name}"
    )

    # Memberikan nama sumbu
    plt.xlabel("Data Points")
    plt.ylabel("Distance")

    # Menampilkan grid
    plt.grid(True)

    # Menampilkan grafik
    plt.show()
```

## Analisis Perbandingan Dendrogram

Berdasarkan hasil percobaan, terdapat perbedaan proses penggabungan
cluster pada metode single linkage, complete linkage, dan average linkage.

Single linkage menentukan jarak antar-cluster berdasarkan jarak terdekat
antara anggota dari dua cluster. Complete linkage menentukan jarak
berdasarkan jarak terjauh, sedangkan average linkage menggunakan
rata-rata jarak antar anggota dari dua cluster.

Pada dataset sederhana, titik d dan f memiliki jarak paling kecil,
yaitu sekitar 0,5, sehingga keduanya cenderung bergabung pada tahap awal.
Titik a dan b juga memiliki jarak yang relatif kecil, yaitu sekitar 0,707.

Perbedaan metode linkage menyebabkan tinggi penggabungan pada dendrogram
berbeda. Single linkage lebih mempertimbangkan pasangan titik yang paling
dekat, complete linkage mempertimbangkan pasangan yang paling jauh,
sedangkan average linkage menggunakan nilai rata-rata jarak.

Dengan demikian, pemilihan metode linkage dapat memengaruhi struktur
cluster yang dihasilkan oleh Hierarchical Clustering.

```python
# ============================================================
# CELL 11 - MEMBACA DATASET MALL CUSTOMERS
# ============================================================

# Membaca dataset Mall Customers dari Google Colab
df = pd.read_csv("/content/Mall_Customers.csv")

# Menampilkan lima baris pertama dataset
print("=== Lima Data Pertama ===")
display(df.head())
```

```python
# ============================================================
# CELL 12 - INFORMASI DATASET
# ============================================================

# Menampilkan ukuran dataset
print("Ukuran dataset:", df.shape)

# Menampilkan informasi tipe data setiap kolom
print("\n=== Informasi Dataset ===")
df.info()
```

```python
# ============================================================
# CELL 13 - MEMILIH FITUR
# ============================================================

# Mengambil kolom ke-4 dan ke-5:
# Annual Income (k$)
# Spending Score (1-100)

x = df.iloc[:, [3, 4]].values

# Menampilkan lima data pertama dari fitur yang dipilih
print("=== Fitur yang Digunakan ===")
print(x[:5])
```

```python
# ============================================================
# CELL 14 - STANDARDISASI DATA
# ============================================================

# StandardScaler digunakan untuk melakukan standardisasi data
from sklearn.preprocessing import StandardScaler

# Membuat objek StandardScaler
data_scaler = StandardScaler()

# Melakukan standardisasi terhadap fitur yang dipilih
scaled_data = data_scaler.fit_transform(x)

# Menampilkan ukuran data setelah standardisasi
print("Ukuran scaled data:", scaled_data.shape)

# Menampilkan beberapa data hasil standardisasi
print("\n=== Lima Data Setelah Standardisasi ===")
print(scaled_data[:5])
```

```python
# ============================================================
# CELL 15 - HIERARCHICAL CLUSTERING
# MALL CUSTOMERS
# ============================================================

# Membuat hierarchical clustering menggunakan:
# - Complete linkage
# - Euclidean distance

clustering = linkage(
    scaled_data,
    method="complete",
    metric="euclidean"
)

# Menampilkan hasil clustering
print("=== Hasil Hierarchical Clustering ===")
print(clustering)
```

```python
# ============================================================
# CELL 16 - DENDROGRAM MALL CUSTOMERS
# ============================================================

# Membuat ukuran grafik
plt.figure(figsize=(12, 8))

# Membuat dendrogram berdasarkan hasil hierarchical clustering
dendrogram(clustering)

# Memberikan judul grafik
plt.title("Hierarchical Clustering Dendrogram - Mall Customers")

# Memberikan nama sumbu
plt.xlabel("Customers")
plt.ylabel("Distance")

# Menampilkan grid
plt.grid(True)

# Menampilkan grafik
plt.show()
```
