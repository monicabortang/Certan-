1. Buatlah sebuah matriks berukuran 7x7 dengan elemen random. Perlihatkan nilai maksimum,nilai minimum, dan nilai rata-rata dari
elemen matriks tersebut.

```python
import numpy as np

print("Tugas 1: Matriks 7x7 dengan Bilangan Bulat Random")
# Membuat matriks 5x5 dengan elemen random
matriks = np.random.randint(1, 101, size=(5, 5))

# Menghitung nilai maksimum, minimum, dan rata-rata
nilai_max = matriks.max()
nilai_min = matriks.min()
nilai_mean = matriks.mean()

print("Matriks 5x5:")
print(matriks)
print("\nNilai Maksimum:", nilai_max)
print("Nilai Minimum :", nilai_min)
print("Nilai Rata-rata:", nilai_mean)
```

2. Buatlah grafik 2 dimensi:

   x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1

```python
import numpy as np
import matplotlib.pyplot as plt

print("Tugas 2: Grafik 2 dimensi")

# Menyiapkan grid x dan y
x = np.linspace(-1.5, 1.5, 500)
y = np.linspace(-1.5, 1.5, 500)
X, Y = np.meshgrid(x, y)

# Persamaan implicit F(x,y) = 1
F = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2

# Plot grafik 2D
plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, levels=[1], colors='red', linewidths=2)

plt.title(r'Grafik 2D: Bentuk Hati')
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

3. Buatlah grafik 3 dimensi 

```python
import numpy as np
import matplotlib.pyplot as plt

print("Tugas 3: Grafik 3 dimensi")

# Menyiapkan data grid x dan y
x = np.linspace(-3, 3, 200)
y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(x, y)

# Menghitung persamaan Z
Z = ((5 * Y / 4) - np.abs(X))**2 + X**2

# Plot grafik 3D surface
fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='plasma', edgecolor='none')

ax.set_title(r'Grafik 3D: $Z = (\frac{5y}{4} - |x|)^2 + x^2$')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```
